```python
import polars as pl
```

```python
df_mstr = pl.read_csv("../data/S&P 500 Historical Components & Changes(03-10-2025).csv")
```

```python
# Ensure date is parsed as a date type
df_mstr = df_mstr.with_columns(
    pl.col("date").str.strptime(pl.Date, "%Y-%m-%d")
)

# Create a complete date range from min to max date
date_range = pl.date_range(
    df_mstr["date"].min(),
    df_mstr["date"].max(),
    "1d",
    eager=True
).alias("date")

```

```python
date_range_df = pl.DataFrame({"date": date_range})

```

```python
# Join with the full date range to fill in missing days
df = date_range_df.join(df_mstr, on="date", how="left")

```

```python
# Forward fill missing values for all columns except 'date'
df = df.sort("date").fill_null(strategy="forward")
```
